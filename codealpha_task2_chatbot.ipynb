```python
import datetime

def get_bot_response(user_input):
    user_input = user_input.lower().strip()

    # Rule-Based Intent Mapping
    if user_input in ["hello", "hi", "hey", "greetings"]:
        return "Hi there! How can I assist you today?"

    elif "how are you" in user_input:
        return "I'm doing great, thanks for asking! How are you doing?"

    elif "your name" in user_input or "who are you" in user_input:
        return "I am AlphaBot, a basic rule-based assistant created for CodeAlpha tasks."

    elif "time" in user_input:
        current_time = datetime.datetime.now().strftime("%I:%M %p")
        return f"The current system time is {current_time}."

    elif "date" in user_input:
        current_date = datetime.datetime.now().strftime("%B %d, %Y")
        return f"Today's date is {current_date}."

    elif user_input in ["bye", "goodbye", "exit", "quit"]:
        return "Goodbye! Have a productive day ahead!"

    else:
        return "I'm sorry, I didn't quite understand that. Try asking about my name, time, or say 'hello'."

def run_chatbot():
    print("=" * 45)
    print("        WELCOME TO ALPHABOT (TASK 4)        ")
    print("   Type 'bye' or 'exit' to terminate chat.  ")
    print("=" * 45)

    while True:
        user_msg = input("\nYou: ")
        response = get_bot_response(user_msg)
        print(f"AlphaBot: {response}")

        if user_msg.lower().strip() in ["bye", "goodbye", "exit", "quit"]:
            break

if __name__ == "__main__":
    run_chatbot()
```
